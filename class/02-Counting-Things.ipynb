## Number of Possible Playlists

Consider a DJ who has a set of $m_s$ salsa songs, $m_b$ bachata songs and $m_r$ reggaeton songs. The DJ plans to play a set of three songs, each of a different genre: first a salsa, then a bachata, and finally a reggaeton song. How many possible sequences of songs can the DJ play?

- There are $m_s$ ways to select a salsa song.
- There are $m_b$ ways to select a bachata song.
- There are $m_r$ ways to select a reggaeton song.

```python
#ex-class

# An example: all possible 3-song playlists where the first sort is a salsa song, the second a bachata song, and the third a reggaeton song.

m_s = 200 # Salsa songs
m_b = 80 # Bachata songs
m_r = 100 # Reggaeton songs

m_s * m_b * m_r
```

```python
#ex-class

salsa = ["Persona ideal (s)", "Que Siga el Baile (s)"]
bachata = ["A la Tierra (b)", "Stand by Me (b)", "Obsesion (b)"]
reggaeton = ["Despacito (r)", "Gasolina (r)"]

playlists = []
for salsa_song in salsa:
    for bachata_song in bachata:
        for reggaeton_song in reggaeton:
            playlists.append((salsa_song, bachata_song, reggaeton_song))

for index, playlist in enumerate(playlists, 1):
    print(f"{index}: {playlist}")

print("Total playlists:", len(playlists))
```

```python
#ex-class

n_values = np.arange(1, 21)
factorial_values = np.array([math.factorial(n) for n in n_values])
exponential_values = np.exp(n_values)

plt.figure(figsize=(10, 6))
plt.plot(n_values, factorial_values, label="n!", marker="o")
plt.plot(n_values, exponential_values, label=r"$e^n$", marker="x")
plt.yscale("log")
plt.title(r"Growth of n! vs. $e^n$")
plt.xlabel("n")
plt.ylabel("Value (log scale)")
plt.legend()
plt.grid(True)
plt.show()
```

## Number of Edges in a Complete Graph

Look at @fig-complete-graph and count the number of edges (lines connecting the vertices).
How many edges are there in a complete graph with $n=1, 2, 3, 10$ vertices?

:::

```python
#ex-class

#| label: fig-complete-graph
#| fig-cap: "Complete graph with 6 nodes"

node_count = 6
angles = np.linspace(0, 2 * np.pi, node_count, endpoint=False)
node_positions = np.column_stack((np.cos(angles), np.sin(angles)))

plt.figure(figsize=(6, 6))

# Draw every edge between distinct nodes
for first_node in range(node_count):
    for second_node in range(first_node + 1, node_count):
        plt.plot(
            node_positions[[first_node, second_node], 0],
            node_positions[[first_node, second_node], 1],
            color="gray",
            linewidth=1,
            zorder=1,
        )

plt.scatter(node_positions[:, 0], node_positions[:, 1], s=500, color="skyblue", zorder=2)

for node_index, (x_position, y_position) in enumerate(node_positions):
    plt.text(x_position, y_position, str(node_index + 1), ha="center", va="center", zorder=3)

plt.axis("equal")
plt.axis("off")
plt.title("Complete graph with 6 nodes")
plt.show()
```

## Birthdays in a Classroom

In our class of (let's say) $n$ students, what is the probability (as a function of $n$) that at least two students share the same birthday? For simplicity, assume that there are 365 days in a year and that each student is equally likely to be born on any day of the year. What is the smallest value of $n$ for which the probability is greater than 1/2?

```python
#ex-class

# Probability of no shared birthdays in a group of 20 people

p_no_shared = (math.factorial(365) / math.factorial(365 - 20)) / (365 ** 20)
print("The probability of no shared birthdays in a group of 20 people is approximately", p_no_shared)

# Probability of at least one shared birthday in a group of 20 people

p_shared = 1 - p_no_shared
print("The probability of at least one shared birthday in a group of 20 people is approximately", p_shared)
```

```python
#ex-class

# Simulation
# The following simulation creates `n_groups` groups of `group_size` people and assigns them
# a random number between 1 and 365 (representing the day of the year).
# We then check if there are any shared birthdays in each group.

n_groups = 1000
group_size = 20

# We select one number between 1 and 365 for each person in each group. The reason we set the upper limit to 366 is that this boundary is not included in the random selection,
# so the last number we can get is 365.

groups = np.random.randint(1, 366, (n_groups, group_size))
groups
```
