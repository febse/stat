## De Morgan's Laws

- The first set is $A = \{1, 2, 3, 4\}$
- The second set is $B = \{2, 5, 6\}$

- Write down the complements of $A$ and $B$ and the intersection of $A$ and $B$. 
- Then, write down the union of $A$ and $B$ and its complement. Compare the results.
