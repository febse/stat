## Two sets

:::{#exm-two-sets}
In a class, 20 students play football, 15 play chess and 6 do both. How many students play at least one of the two games?

$$
20 + 15 - 6 = 29
$$
:::

```python
football = set(range(1, 21))      # students 1, ..., 20
chess = set(range(15, 30))        # students 15, ..., 29 (15 students)

print("|A| + |B| - |A ∩ B| =", len(football) + len(chess) - len(football & chess))
print("|A ∪ B|             =", len(football | chess))
```

## Three sets

```python
A = {n for n in range(1, 101) if n % 2 == 0}   # multiples of 2
B = {n for n in range(1, 101) if n % 3 == 0}   # multiples of 3
C = {n for n in range(1, 101) if n % 5 == 0}   # multiples of 5

by_formula = (
    len(A) + len(B) + len(C)
    - len(A & B) - len(A & C) - len(B & C)
    + len(A & B & C)
)

print("By the formula:", by_formula)
print("Direct count:  ", len(A | B | C))
```

## The general formula

:::{#thm-inclusion-exclusion}
## Inclusion/exclusion

For finite sets $A_1, \ldots, A_n$,

$$
\left|\bigcup_{i=1}^{n} A_i\right| = \sum_{k=1}^{n} (-1)^{k+1} \sum_{1 \le i_1 < \ldots < i_k \le n} \left|A_{i_1} \cap \ldots \cap A_{i_k}\right|.
$$

In words: add the sizes of all single sets, subtract the sizes of all pairwise intersections, add the sizes of all triple intersections, and so on, alternating signs.
:::

:::{.callout-note collapse="true"}
## Proof (click to expand)

Take any element $x$ of the union and let $m \ge 1$ be the number of sets that contain $x$. We show that the right-hand side counts $x$ exactly once.

The element $x$ lies in exactly $\binom{m}{k}$ of the $k$-fold intersections (we choose which $k$ of its $m$ sets form the intersection). Its total contribution is therefore

$$
\sum_{k=1}^{m} (-1)^{k+1} \binom{m}{k} = 1 - \sum_{k=0}^{m} (-1)^{k} \binom{m}{k} = 1 - 0 = 1,
$$

because $\sum_{k=0}^{m} (-1)^k \binom{m}{k} = (1 - 1)^m = 0$ for $m \ge 1$ (binomial theorem). Elements outside the union appear in none of the sets and contribute $0$. $\blacksquare$
:::

```python
for m in range(1, 8):
    contribution = sum((-1) ** (k + 1) * math.comb(m, k) for k in range(1, m + 1))
    print(f"element in m = {m} sets is counted {contribution} time(s)")
```

## Exercises

How many integers from 1 to 1000 are divisible by at least one of 3, 5 or 7?

Use the three-set formula and verify your answer with code.

:::{.callout-note collapse="true"}
## Solution (click to expand)

Let $A$, $B$, $C$ be the multiples of 3, 5 and 7 in $\{1, \ldots, 1000\}$. A number is divisible by two (three) of them exactly when it is divisible by their product, so $|A \cap B|$ is the number of multiples of 15, and so on.

$$
\begin{aligned}
|A| &= \lfloor 1000/3 \rfloor = 333, & |B| &= \lfloor 1000/5 \rfloor = 200, & |C| &= \lfloor 1000/7 \rfloor = 142,\\
|A \cap B| &= \lfloor 1000/15 \rfloor = 66, & |A \cap C| &= \lfloor 1000/21 \rfloor = 47, & |B \cap C| &= \lfloor 1000/35 \rfloor = 28,\\
|A \cap B \cap C| &= \lfloor 1000/105 \rfloor = 9.
\end{aligned}
$$

$$
|A \cup B \cup C| = 333 + 200 + 142 - 66 - 47 - 28 + 9 = 543.
$$
:::

```python
by_formula = (1000 // 3 + 1000 // 5 + 1000 // 7
              - 1000 // 15 - 1000 // 21 - 1000 // 35
              + 1000 // 105)
direct = sum(1 for number in range(1, 1001) if number % 3 == 0 or number % 5 == 0 or number % 7 == 0)

print(by_formula, direct)
```
