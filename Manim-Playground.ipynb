What is a nice video presentation?

- It should be no longer than necessary but it should also explain the problem
- It should not contain anything unrelated
- Keep the number of visual cues (colors, shapes, etc.) to the bare minimum necessary
- Focuse on the formulas, use as little free text as possible

### Render a test video using Manim

```python
from manim import *

class MathTestScene(Scene):
    def construct(self):
        # Math equation using LaTeX
        equation = MathTex(r"e^{i\pi} = -1")
        equation.scale(2.0)

        self.play(Write(equation))
        self.wait(1)
        self.play(FadeOut(equation))

# Run the render command
%manim -v WARNING -qm MathTestScene
```
