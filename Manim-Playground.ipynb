### Render a test video using Manim

```python
from manim import *

class MathTestScene(Scene):
    def construct(self):
        # Math equation using LaTeX
        equation = MathTex(r"e^{i\pi} + 1 = 0")
        equation.scale(2.0)

        self.play(Write(equation))
        self.wait(1)
        self.play(FadeOut(equation))

# Run the render command
%manim -v WARNING -qm MathTestScene
```
