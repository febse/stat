In the previous section (@sec-discrete-random-variables) we introduced the concept of a random variable that can take
values among a finite set ot outcomes, for example yes/no, number of successes/failures, etc which we called discrete variables.

Now let's turn our attention to sums of such variables. For example, consider a sequence of games. You throw a dice
two times (independently) and win the sum of the dice. Let's write $X_1$ for the first game, $X_2$ for the second one,
then their sum will be 

$$
S_2 = X_1 + X_2
$$

The sum of three consecutive throws will be 

$$
S_3 = X_1 + X_2 + X_3 = S_2 + X_3
$$

Let's assume that the die is fair so each of its four sides has a probability of $1/4$. Can we derive the distribution
of these sums from the underlying distribution of the dice throws? It turns out that there is a pretty elegant way to do so, 
called the *convolution* of the distributions of the dice.

## The Distribution of a Sum of Random Variables

Let's start with $S_2$. We will call the probability mass functions of $X_1$ and $X_2$ $f_{X_1}$ and $f_{X_2}$.

$$
f_{X_1}(x) = \begin{cases}
1/4 & x=0 \\
1/4 & x=1 \\
1/4 & x=2 \\
1/4 & x=3 \\
1/4 & x=4 \\
0 & \text{otherwise}
\end{cases}
\quad 
f_{X_2}(x) = \begin{cases}
1/4 & x=0 \\
1/4 & x=1 \\
1/4 & x=2 \\
1/4 & x=3 \\
1/4 & x=4 \\
0 & \text{otherwise}
\end{cases}
$$


The probability of their sum is given by the convolution formula

$$
\underbrace{f_{S_2}(s)}_{\text{Prob. } S_2=s} = \sum_{k} f_{X_1}(k)f_{X_2}(s - k)
$$

Let's decypher what it tells us. The left hand side is the probability mass function of the sum, we call it $f_{S_2}$. You can call it anything, e.g. $h(i)$ as long as you remember what it means. Its argument $s$ is a concrete value, one of $\{2, 3, 4, 5, \ldots, 8\}$ because the sum of two 4-sided dice cannot be less than 2 or more than 8. Let's start with $s=2$. Remember that the PMF function must return the probability of its argument.

$$
f_{S_2}(2) = \sum_{k} f_{X_1}(k)f_{X_2}(2 - k)
$$

Now let's read what we have. $\sum_{k}$ is a compact way to say: sum over all possible values of $k$ which are $\{1, 2, 3, 4\}$ so let's do this:

$$
\begin{align*}
f_{S_2}(2) & = \underbrace{f_{X_1}(1)f_{X_2}(2 - 1)}_{k=1} \\
           & + \underbrace{f_{X_1}(2)f_{X_2}(2 - 2)}_{k=2} \\
           & + \underbrace{f_{X_1}(3)f_{X_2}(2 - 3)}_{k=3} \\ 
           & + \underbrace{f_{X_1}(4)f_{X_2}(2 - 4)}_{k=4}  \\
           & = \frac{1}{4} \cdot \frac{1}{4}
\end{align*}
$$

In this case all term except the first one are zero. Why (try to answer yourself before seeing the answer)?

:::{.callout-note collapse="true"}

$$
f_{X_1}(2)f_{X_2}(2 - 2) = f_{X_1}(2)f_{X_2}(0) = 0
$$

This is because of the second PMF: $f_{X_2}(0) = 0$. Look at its definition above, this is where the "otherwise" bites.
The rest of the terms are zero by the same logic.

:::

So now we found that $f_{S_2} = 1/16$. Let's try it with $s=3$.

$$
\begin{align*}
f_{S_2}(3) & = \underbrace{f_{X_1}(1)f_{X_2}(3 - 1)}_{k=1} \\
           & + \underbrace{f_{X_1}(2)f_{X_2}(3 - 2)}_{k=2} \\
           & + \underbrace{f_{X_1}(3)f_{X_2}(3 - 3)}_{k=3} \\ 
           & + \underbrace{f_{X_1}(4)f_{X_2}(3 - 4)}_{k=4}  \\
           & = f_{X_1}(1)f_{X_2}(2) + f_{X_1}(2)f_{X_2}(1) \\
           & = \frac{1}{4} \cdot \frac{1}{4} + \frac{1}{4} \cdot \frac{1}{4} \\
           & = \frac{2}{16}
\end{align*}
$$

Now stop and look at the pattern. The formula gives us the sum of the probabilities of outcomes of the first die and the second die
for which the *sum of the outcomes* is one and the same value: 3 at the current step.

Continue writing the sum for $s=4$ and $s=5$ and calculate the resulting probability as an exercise.
