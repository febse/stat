# Fallacies With Conditional Probabilities

Bayes' theorem is often used in court to decide how much a piece of evidence should change our belief in a verdict. Most mistakes come from mixing up two conditional probabilities: the probability of the evidence given the verdict, $P(E \mid G)$, and the probability of the verdict given the evidence, $P(G \mid E)$. In this chapter we look at a real example of such a mistake, the _defender's fallacy_.

## The Defender's Fallacy: Abuse and Murder

::: {#exr-defender-fallacy .ex-class}
## Abuse and Murder

In the 1995 murder trial of O. J. Simpson, the defense argued that a history of abuse was irrelevant to the case because "only 1 in 2,500 abused women is murdered by her abuser" [@gigerenzer2002reckoning]. The same point was made by @good1995batterer. In this exercise we examine this argument.

Consider a woman who was abused by her partner and define the following events:

- $A$: the woman was abused by her partner.
- $M$: the woman was murdered during the period considered (one year).
- $P$: the murderer is her partner (the partner is the murderer).
- $P^c$: the murderer is someone other than her partner.

Of every 100,000 abused women, in one year:

- 40 are murdered by their partner (the event $M \cap P$),
- 5 are murdered by someone else (the event $M \cap P^c$).

1. **The defense's number.** Write the figure $1/2500$ as a probability of the events above. Which population does it average over?
2. **Counting.** Among 100,000 abused women, how many are murdered in total? How many of them were killed by their partner?
3. **Bayes.** Compute $P(P \mid A \cap M)$, the probability that the partner is the murderer given that the woman was abused and murdered. Use the partition $\{P, P^c\}$ of the event $M$.
4. **The flaw.** What did the defense condition on, and what did it leave out? Why is the omitted fact the core of the trial?
5. **Comparison.** The prosecutor's fallacy confuses $P(E \mid G^c)$ with $P(G^c \mid E)$. Is the defender's fallacy of the same kind? Which fallacy inflates and which deflates the probability of guilt?
:::

:::{.callout-note collapse="true"}
## Solution (click to expand)

**1. The defense's number.** The figure $1/2500$ is the probability that an abused woman is murdered _by her partner_ in a given year:

$$
P(M \cap P \mid A) = \frac{40}{100{,}000} = \frac{1}{2500}
$$

It averages over _all_ abused women, and almost all of them are alive. It is not the probability we need.

**2. Counting.** Out of 100,000 abused women, $40 + 5 = 45$ are murdered, and 40 of these 45 were killed by their partner.

**3. Bayes.** The events $M \cap P$ and $M \cap P^c$ partition the event $M$. By the definition of conditional probability and the total probability theorem,

$$
P(P \mid A \cap M) = \frac{P(M \cap P \mid A)}{P(M \mid A)} = \frac{P(M \cap P \mid A)}{P(M \cap P \mid A) + P(M \cap P^c \mid A)} = \frac{40/100{,}000}{40/100{,}000 + 5/100{,}000} = \frac{40}{45} = \frac{8}{9} \approx 0.89
$$

This agrees with the value of about 8 in 9 reported by @gigerenzer2002reckoning.

**4. The flaw.** The defense conditioned on abuse ($A$) only. It left out the fact that the woman _was murdered_ ($M$). Once $M$ is known, the 99,955 abused women who were not murdered are irrelevant, and the question is who among the murderers is the partner. The probability of the partner being the murderer is then about $8/9$ and not $1/2500$.

**5. Comparison.** Both fallacies condition on the wrong event, but in opposite directions. The prosecutor's fallacy replaces $P(G^c \mid E)$ by the small number $P(E \mid G^c)$ and so _inflates_ the probability of guilt. The defender's fallacy uses a small probability that does not condition on all the known facts ($M$ is missing) and so _deflates_ it.

**Robustness.** The figure of 5 murders per 100,000 by someone else is the least well documented number. Even if it were three times as large (15 per 100,000), we would get $40/55 \approx 0.73$, which is still far from $1/2500$.
:::

We can check the counting argument with a short calculation.

```python
abused_women = 100_000
murdered_by_partner = 40
murdered_by_someone_else = 5

murdered_total = murdered_by_partner + murdered_by_someone_else

print(f"P(murdered by partner | abused)         = {murdered_by_partner / abused_women:.6f}  (1/2500 = {1 / 2500:.6f})")
print(f"P(partner is murderer | abused, murdered) = {murdered_by_partner / murdered_total:.4f}  (8/9 = {8 / 9:.4f})")
```
